```python
import json
import os

import pandas as pd
import torch

os.chdir("..")
os.getcwd()
```

```python
labelled = True

output_dir = "/Users/gabriele/Desktop/aether/data/s2bms/aux_stats/"
os.makedirs(output_dir, exist_ok=True)

if labelled:
    df = pd.read_csv("/Users/gabriele/Desktop/aether/data/s2bms/model_ready_s2bms.csv")
    output_path = output_dir + "labelled.json"
else:
    df = pd.read_csv(
        "/Users/gabriele/Desktop/aether/data/s2bms/model_ready_s2bms-unlabelled-20260529.csv"
    )
    output_path = output_dir + "unlabelled.json"

col = [i for i in df.columns if "aux" in i and "top" not in i]
```

```python
# Load in original datasplit file
if labelled:
    pth = "data/s2bms/splits/s2bms_union_val_test.pth"
else:
    pth = "data/s2bms/splits/s2bms_unlabelled_union_val_test.pth"


split_indices = torch.load(pth, weights_only=False)
```

```python
df = df[df["name_loc"].isin(split_indices["train_indices"])]
```

```python
if os.path.exists(output_path):
    print(f"Output file already exists: {output_path}")
else:
    stats = {}
    for col in col:
        stats[col] = {
            "mean": float(df[col].mean()),
            "std": float(df[col].std()),
        }

        with open(output_path, "w") as f:
            json.dump(stats, f, indent=4)

        print(f"Saved global statistics for {len(stats)} variables to {output_path}")
```
